**Título:** Decidibilidade, o Problema da Parada de Turing e Classes de Complexidade P vs NP

> **Enunciado Original:**
> Demonstrar formalmente por redução e diagonalização de Cantor a indecidibilidade do Problema da Parada (Halting Problem) e situar as classes P, NP, NP-Completo e NP-Difícil.

### Formulação Matemática Formal
$$H(\langle M, w \rangle) = \begin{cases} \text{Sim}, & \text{se } M(w) \text{ para} \\ \text{Não}, & \text{se } M(w) \text{ entra em loop} \end{cases} \implies \text{Paradoxo Lógico}$$


### Passo 1: Suposição da Existência do Decisor Universal H
Supor que existe máquina de Turing H capaz de decidir se qualquer M para com entrada w.


### Passo 2: Construção da Máquina Paradoxal D
Construir D que toma a descrição \langle M \rangle como entrada: chama H(\langle M, \langle M \rangle \rangle) e inverte o resultado (entra em loop se H responder Sim; para se H responder Não).


### Passo 3: Execução de D com sua Própria Descrição
O que acontece com D(\langle D \rangle)? Se D para, H diz que para, logo D entra em loop (Contradição!). Se D entra em loop, H diz que não para, logo D para (Contradição!).


### Passo 4: Conclusão da Indecidibilidade
Nenhuma máquina H pode existir. O Problema da Parada é categoricamente indecidível.


### Passo 5: Relação com Classes P e NP
Problemas verificáveis em tempo polinomial (NP) vs decidíveis em tempo polinomial (P) com problemas NP-Completos sendo os mais difíceis de NP.


```python
import numpy as np
import matplotlib.pyplot as plt

classes = ['P (Polinomial)', 'NP (Verificável)', 'NP-Completo (3-SAT)', 'Indecidível (Parada)']
complexidade = [1, 3, 4, 10]
cores = ['#10b981', '#38bdf8', '#f59e0b', '#ef4444']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211a09')
ax.set_facecolor('#211a09')
bars = ax.barh(classes, complexidade, color=cores, height=0.45)
ax.set_title('Hierarquia de Computabilidade e Complexidade', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Grau Relativo de Intratabilidade / Dificuldade', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='x', color='#4d3d15', alpha=0.5)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
