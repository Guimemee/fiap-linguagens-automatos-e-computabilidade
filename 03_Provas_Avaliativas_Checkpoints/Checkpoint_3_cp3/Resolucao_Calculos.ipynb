> **Enunciado Original:**
> Utilizar o Lema do Bombeamento (Pumping Lemma) para provar por contradição que a linguagem L = {a^n b^n | n >= 0} não é regular e não pode ser reconhecida por nenhum AFD.

### Formulação Matemática Formal
$$\forall w \in L, |w| \ge p \implies w = xyz, \quad |xy| \le p, \quad |y| \ge 1, \quad \forall i \ge 0: x y^i z \in L$$


### Passo 1: Hipótese por Contradição
Supor que L = \{a^n b^n \mid n \ge 0\} é uma linguagem regular com comprimento de bombeamento p.


### Passo 2: Escolha da Cadeia Crítica
Selecionar a cadeia w = a^p b^p \in L com comprimento |w| = 2p \ge p.


### Passo 3: Decomposição Obrigatória w = xyz
Como |xy| \le p, a subcadeia y consiste exclusivamente de símbolos 'a' (isto é, y = a^k, onde 1 \le k \le p).


### Passo 4: Bombeamento com i = 2
A cadeia bombeada x y^2 z possui p + k símbolos 'a' e apenas p símbolos 'b'.


### Passo 5: Contradição e Conclusão
Como p + k \ne p, x y^2 z \notin L, refutando a hipótese. Logo, L não é regular e exige autômato com pilha.


```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(1, 11)
simbolos_a = n
simbolos_b = n
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211a09')
ax.set_facecolor('#211a09')
ax.plot(n, simbolos_a, 'o-', color='#eab308', lw=2.5, label='Contagem de Símbolos 'a'')
ax.plot(n, simbolos_b, 's--', color='#38bdf8', lw=2, label='Contagem de Símbolos 'b' (Memória Requerida)')
ax.set_title('Incapacidade de Reconhecimento por Memória Finita', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Expoente n da Cadeia aⁿbⁿ', color='#94a3b8')
ax.set_ylabel('Quantidade de Símbolos', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d3d15', alpha=0.5)
ax.legend(facecolor='#33280e', edgecolor='#eab308', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
