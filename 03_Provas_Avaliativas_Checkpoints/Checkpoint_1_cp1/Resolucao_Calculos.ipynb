> Demonstrar formalmente as propriedades de fechamento de linguagens regulares sob operações de união, concatenação, fecho de Kleene, interseção e complemento.

### Formulação Matemática Formal
$$L_1 \cup L_2 = \{w \mid w \in L_1 \lor w \in L_2\}, \quad L^* = \bigcup_{i=0}^\infty L^i, \quad L^0 = \{\epsilon\}$$


### Passo 1: Definição Formal de Alfabeto e Cadeia
\Sigma = \{0, 1\}; cadeias w \in \Sigma^* com comprimento |w| \ge 0.


### Passo 2: Fechamento sob União e Concatenação
Construção com autômato AFN com transições-\epsilon provando que a classe de linguagens regulares é fechada sob união.


### Passo 3: Fechamento sob Complemento
Inversão dos estados de aceitação F' = Q \setminus F em AFD determinístico completo.


### Passo 4: Leis de De Morgan para Interseção
L_1 \cap L_2 = \overline{\overline{L_1} \cup \overline{L_2}}, provando fechamento sob interseção.


### Passo 5: Conclusão Teórica
A álgebra de linguagens regulares forma uma álgebra booleana fechada completa.


```python
import numpy as np
import matplotlib.pyplot as plt

operacoes = ['União', 'Interseção', 'Complemento', 'Concatenação', 'Fecho Kleene']
fechamento = [1, 1, 1, 1, 1]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211a09')
ax.set_facecolor('#211a09')
bars = ax.bar(operacoes, fechamento, color='#eab308', width=0.45)
ax.set_ylim(0, 1.3)
ax.set_title('Fechamento Algébrico das Linguagens Regulares', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Propriedade Válida (1 = Sim)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#4d3d15', alpha=0.5)
for bar in bars:
    ax.text(bar.get_x() + bar.get_width()/2.0, 1.05, 'FECHADO', ha='center', va='bottom', color='#ffffff', fontweight='bold', fontsize=9)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
