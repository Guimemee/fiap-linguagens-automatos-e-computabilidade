> Projetar a Gramática Livre de Contexto em Forma Normal de Chomsky (FNC) e o Autômato com Pilha determinístico para reconhecer expressões aritméticas balanceadas com parênteses.

### Formulação Matemática Formal
$$A \to BC \lor A \to a \quad (\text{Forma Normal de Chomsky}), \quad \delta: Q \times (\Sigma \cup \{\epsilon\}) \times \Gamma \to \mathcal{P}(Q \times \Gamma^*)$$


### Passo 1: Regras de Produção da GLC
S \to S + S \mid S * S \mid (S) \mid a.


### Passo 2: Eliminação de Ambiguidade
Separação em níveis de precedência: E (Expressão/Soma), T (Termo/Multiplicação), F (Fator/Parênteses).


### Passo 3: Conversão para Forma Normal de Chomsky
Eliminação de produções-\epsilon, produções unitárias e substituição por produções estritamente binárias.


### Passo 4: Operação da Pilha do Autômato
Empilhamento de '(' ao encontrar abre-parênteses e desempilhamento correspondente em fecha-parênteses ')'.


### Passo 5: Conclusão de Reconhecimento
Aceitação por pilha vazia e estado final garantindo parsing sintático de compiladores em O(N^3).


```python
import numpy as np
import matplotlib.pyplot as plt

profundidade = np.arange(1, 9)
elementos_pilha = profundidade
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211a09')
ax.set_facecolor('#211a09')
ax.plot(profundidade, elementos_pilha, 's-', color='#eab308', lw=2.5, label='Ocupação da Pilha (Memória LIFO)')
ax.set_title('Evolução da Pilha com Parênteses Aninhados', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Profundidade de Aninhamento de Parênteses', color='#94a3b8')
ax.set_ylabel('Símbolos na Pilha Γ', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d3d15', alpha=0.5)
ax.legend(facecolor='#33280e', edgecolor='#eab308', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
