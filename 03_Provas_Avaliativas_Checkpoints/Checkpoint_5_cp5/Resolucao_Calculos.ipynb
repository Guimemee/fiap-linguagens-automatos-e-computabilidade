> **Enunciado Original:**
> Construir a função de transição de uma Máquina de Turing padrão de fita infinita para reconhecer a linguagem de cópia L = {w#w | w in {0,1}*} e formalizar a Tese de Church-Turing.

### Formulação Matemática Formal
$$M = (Q, \Sigma, \Gamma, \delta, q_0, B, F), \quad \delta(q, X) = (p, Y, D), \quad D \in \{L, R\}$$


### Passo 1: Estrutura da Fita e Cabeçote de Leitura/Gravação
Alfabeto de fita \Gamma = \{0, 1, X, Y, \#, B\} com símbolo em branco B.


### Passo 2: Algoritmo de Casamento de Símbolos
O cabeçote lê o primeiro símbolo antes de '#', substitui por X, move à direita até a posição correspondente após '#' e valida.


### Passo 3: Retorno do Cabeçote à Esquerda
Retorno à esquerda até o próximo símbolo não marcado.


### Passo 4: Verificação de Término
Quando todos os símbolos à esquerda de '#' forem consumidos, verifica se não restam símbolos residuais à direita.


### Passo 5: Conclusão de Computabilidade
A máquina atinge o estado de aceitação q_{accept} em O(N^2) passos, confirmando que L é uma linguagem recursiva (decidível).


```python
import numpy as np
import matplotlib.pyplot as plt

n_tamanho = np.array([2, 4, 8, 16, 32])
passos_turing = 2 * n_tamanho**2 + 5 * n_tamanho
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211a09')
ax.set_facecolor('#211a09')
ax.plot(n_tamanho, passos_turing, 'o-', color='#f59e0b', lw=2.5, label='Transições da Fita (Passos)')
ax.set_title('Complexidade Temporal da Máquina de Turing (w#w)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tamanho da Cadeia w (Caracteres)', color='#94a3b8')
ax.set_ylabel('Passos de Execução do Cabeçote', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d3d15', alpha=0.5)
ax.legend(facecolor='#33280e', edgecolor='#eab308', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
