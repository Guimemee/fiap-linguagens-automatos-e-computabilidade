**Título:** Autômatos Finitos Determinísticos (AFD) e Minimização de Hopcroft

> **Enunciado Original:**
> Construir o AFD M = (Q, Sigma, delta, q0, F) que reconhece cadeias binárias com número par de 0s e número ímpar de 1s e aplicar o algoritmo de minimização de estados de Hopcroft.

### Formulação Matemática Formal
$$M = (Q, \Sigma, \delta, q_0, F), \quad \delta: Q \times \Sigma \to Q, \quad |Q_{min}| = 4$$


### Passo 1: Espaço de Estados do Produto Cartesiano
Estados Q = \{q_{par,par}, q_{par,impar}, q_{impar,par}, q_{impar,impar}\} correspondentes às paridades.


### Passo 2: Função de Transição de Estados delta
Transições com '0' alternam a paridade de zeros; transições com '1' alternam a paridade de uns.


### Passo 3: Estado Inicial e Conjunto de Aceitação
q_0 = q_{par,par}; \quad F = \{q_{par,impar}\}.


### Passo 4: Aplicação da Minimização de Hopcroft
Separação inicial em blocos F e Q \setminus F. Nenhum par de estados é indistinguível.


### Passo 5: Conclusão Estrutural
O AFD obtido já é canonicamente mínimo com exatamente 4 estados e zero redundância.


```python
import numpy as np
import matplotlib.pyplot as plt

estados = ['q_par_par (Início)', 'q_par_impar (Aceita)', 'q_impar_par', 'q_impar_impar']
grau = [2, 2, 2, 2]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211a09')
ax.set_facecolor('#211a09')
bars = ax.bar(estados, grau, color=['#94a3b8', '#10b981', '#f59e0b', '#f59e0b'], width=0.45)
ax.set_title('Grau de Saída por Estado do AFD Canônico', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Transições Determinísticas', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#4d3d15', alpha=0.5)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
